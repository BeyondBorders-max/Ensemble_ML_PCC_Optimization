# NSGA-III Many-Objective Optimization — PCC Ensemble Surrogate
## Step 8: Six-Objective Pareto Optimization | Step 9: Aspen Validation & Knee-Point Selection

**Objectives (6):**
| # | Objective | Direction |
|---|-----------|----------|
| 1 | CO₂ Recovery (%) | Maximize |
| 2 | CO₂ Purity (mole%) | Maximize |
| 3 | SEC Energy (kJ/kg CO₂) | Minimize |
| 4 | Solvent Loss (kg/h) | Minimize |
| 5 | OPEX (USD/t CO₂) | Minimize |
| 6 | Corrosion Rate (mm/yr) | Minimize |

**Algorithm:** NSGA-III | **Surrogates:** Stacked Meta-Surrogate + Bagged CatBoost
**Result:** 132 Aspen-validated Pareto-optimal solutions

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import joblib
import warnings
warnings.filterwarnings('ignore')

try:
    from pymoo.algorithms.moo.nsga3 import NSGA3
    from pymoo.util.ref_dirs import get_reference_directions
    from pymoo.core.problem import Problem
    from pymoo.optimize import minimize
    from pymoo.operators.crossover.sbx import SBX
    from pymoo.operators.mutation.pm import PM
    from pymoo.operators.sampling.rnd import FloatRandomSampling
    PYMOO_AVAILABLE = True
    print(f"\u2713 pymoo installed")
except ImportError:
    PYMOO_AVAILABLE = False
    print("\u26a0 pymoo not found \u2014 pip install pymoo>=0.6")

print(f"NumPy: {np.__version__} | Pandas: {pd.__version__}")
```

## Load Pre-Trained Surrogate Models
The Stacked Meta-Surrogate (R²=0.9956) is used as the primary evaluator.
The Bagged CatBoost (R²=0.9947) provides redundancy checking.

```python
import os

scaler_X  = joblib.load('scaler_X.pkl')
scaler_y  = joblib.load('scaler_y.pkl')
stacked   = joblib.load('08_Stacked_Meta_Surrogate.pkl')
bagged_cb = joblib.load('07_Bagged_CatBoost.pkl')

# Load all base models for stacking meta-input construction
ann = joblib.load('01_ANN_model.pkl') if os.path.exists('01_ANN_model.pkl') else None
pso_catboost = joblib.load('03_PSO_CatBoost.pkl')
pso_gbr      = joblib.load('04_PSO_GBR.pkl')
pso_xgb      = joblib.load('05_PSO_XGBoost.pkl')
pso_ada      = joblib.load('06_PSO_AdaBoost.pkl')

print("\u2713 All surrogate models loaded")

FINAL_FEATURES = [
    'flue_gas_mass_flow', 'flue_gas_co2_mol', 'flue_gas_temp', 'flue_gas_h2o_mol',
    'lean_solvent_mass_flow', 'lean_solvent_temp', 'amine_type', 'amine_wt_pct',
    'absorber_pressure', 'absorber_packing_height', 'desorber_pressure', 'reboiler_temp',
    'reboiler_duty', 'stripper_reflux_ratio', 'stripper_packing_height', 'rich_solvent_temp_in',
    'lean_rich_hx_effectiveness', 'lean_rich_hx_min_approach', 'cooling_water_temp_in',
    'condenser_outlet_temp', 'absorber_pressure_drop', 'stripper_pressure_drop',
    'flue_gas_so2_ppm', 'flue_gas_no2_ppm', 'heat_stable_salts_wt_pct',
    'corrosion_inhibitor_wt_pct', 'solvent_makeup_rate', 'material_category',
    'cooling_water_cost', 'low_pressure_steam_cost'
]

TARGET_COLS = ['co2_recovery', 'co2_purity', 'sec_energy',
               'solvent_loss', 'opex_usd_per_t', 'corrosion_rate']
```

## Define NSGA-III Problem
6-objective problem with 30 decision variables (continuous + categorical).
Reference directions: Das-Dennis method with 7 partitions → 462 reference points.

```python
# Parameter bounds for 30 inputs
BOUNDS_LOWER = np.array([
    5e5, 0.02, 35.0, 3.97, 8e5, 25.0, 1.0, 14.95,
    1.05, 10.0, 1.50, 110.0, 245.0, 0.5, 9.51, 84.66,
    0.72, 5.0, 20.0, 26.06, 5.0, 18.26, 0.0, 0.0, 1.52,
    0.01, 16.78, 1.0, 0.03, 10.0
])
BOUNDS_UPPER = np.array([
    3.6e6, 0.045, 45.0, 10.22, 3.5e6, 60.0, 5.0, 60.0,
    1.30, 21.83, 2.20, 119.81, 510.0, 1.44, 16.75, 97.17,
    0.86, 12.3, 34.33, 45.0, 20.0, 36.31, 50.0, 20.0, 4.35,
    2.0, 50.0, 4.0, 1.5, 15.5
])

class PCCProblem(Problem):
    def __init__(self, surrogates, scaler_X, scaler_y):
        super().__init__(n_var=30, n_obj=6, n_constr=0,
                         xl=BOUNDS_LOWER, xu=BOUNDS_UPPER)
        self.surrogates = surrogates
        self.scaler_X = scaler_X
        self.scaler_y = scaler_y

    def _evaluate(self, X, out, *args, **kwargs):
        X_sc = self.scaler_X.transform(X)
        # Build meta-input from base surrogates
        preds = []
        for m in self.surrogates['base']:
            preds.append(self.scaler_y.inverse_transform(m.predict(X_sc)))
        meta_X = np.hstack(preds)
        y_pred = self.surrogates['meta'].predict(meta_X)

        # Objectives: negate recovery+purity (maximize \u2192 minimize)
        f1 = -y_pred[:, 0]  # -CO2_recovery
        f2 = -y_pred[:, 1]  # -CO2_purity
        f3 =  y_pred[:, 2]  # SEC (minimize)
        f4 =  y_pred[:, 3]  # solvent_loss (minimize)
        f5 =  y_pred[:, 4]  # opex (minimize)
        f6 =  y_pred[:, 5]  # corrosion_rate (minimize)

        out["F"] = np.column_stack([f1, f2, f3, f4, f5, f6])

problem = PCCProblem(
    surrogates={'base': [pso_catboost, pso_gbr, pso_xgb, pso_ada], 'meta': stacked},
    scaler_X=scaler_X, scaler_y=scaler_y
)
print("\u2713 6-objective PCCProblem defined (30 decision variables)")
```

## Run NSGA-III Optimization
Population: 300 | Generations: 500 | Crossover: SBX (η=15) | Mutation: PM (η=20)
Reference directions: Das-Dennis (7 partitions) → 462 reference points

```python
if PYMOO_AVAILABLE:
    ref_dirs = get_reference_directions("das-dennis", 6, n_partitions=7)

    algorithm = NSGA3(
        ref_dirs=ref_dirs,
        pop_size=300,
        sampling=FloatRandomSampling(),
        crossover=SBX(eta=15, prob=0.9),
        mutation=PM(eta=20),
    )

    result = minimize(
        problem, algorithm,
        termination=('n_gen', 500),
        seed=42, verbose=True
    )

    # Extract Pareto front
    X_pareto = result.X
    F_pareto = result.F

    df_pareto = pd.DataFrame(X_pareto, columns=FINAL_FEATURES)
    df_pareto[TARGET_COLS] = F_pareto * np.array([-1, -1, 1, 1, 1, 1])

    print(f"\n\u2713 NSGA-III complete: {len(df_pareto)} Pareto solutions found")
    df_pareto.to_csv('pareto_front_raw.csv', index=False)
else:
    print("Loading pre-generated Pareto solutions...")
    df_pareto_path = 'pareto_front_optimal_solutions.csv' if os.path.exists('pareto_front_optimal_solutions.csv') else '../Ensemble_PCC_Project_Results/NSGA-III/data/pareto_front_optimal_solutions.csv'
    df_pareto = pd.read_csv(df_pareto_path)
    print(f"\u2713 Loaded {len(df_pareto)} Pareto solutions")
```

## Step 9: Knee-Point Selection — Multi-Criteria Decision Making

The **Knee-Point** is defined as the solution with the minimum Euclidean distance 
to the ideal (utopia) point in the normalized objective space. This approach avoids 
the arbitrary weighting of TOPSIS and identifies the solution offering the best 
balanced trade-off across all 6 objectives simultaneously.

**Method:**
1. Normalize all Pareto objectives to [0, 1] range
2. Define ideal point: min of each normalized objective
3. Compute Euclidean distance from each solution to ideal point
4. Select solution with minimum distance (the "knee" of the Pareto front)

```python
df_pf = df_validated.copy()
obj_cols = TARGET_COLS

# Normalize to [0,1]: for maximize objectives (recovery, purity) \u2192 invert
obj_scaled = df_pf[obj_cols].copy()
obj_scaled['co2_recovery'] = -obj_scaled['co2_recovery']  # negate to minimize
obj_scaled['co2_purity']   = -obj_scaled['co2_purity']

# Min-max normalize
for c in obj_scaled.columns:
    mn, mx = obj_scaled[c].min(), obj_scaled[c].max()
    obj_scaled[c] = (obj_scaled[c] - mn) / (mx - mn + 1e-10)

# Ideal point: 0 for all (after normalization of minimization objectives)
ideal = np.zeros(6)
distances = np.sqrt(((obj_scaled.values - ideal)**2).sum(axis=1))
knee_idx = np.argmin(distances)

df_knee = df_pf.iloc[[knee_idx]].copy()
df_knee['Knee_Distance'] = distances[knee_idx]
print("=== KNEE-POINT OPTIMAL SOLUTION ===")
for t in TARGET_COLS:
    if t in df_knee.columns:
        print(f"  {t}: {df_knee[t].values[0]:.4f}")

df_knee.to_csv('kneepoint_optimal_solution.csv', index=False)
print(f"\nKnee distance: {distances[knee_idx]:.4f}")
```
